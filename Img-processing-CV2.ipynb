# Morphological operators

```python
def load_img():
    blank_img = np.zeros((600,600))
    font = cv2.FONT_HERSHEY_SIMPLEX
    cv2.putText(blank_img, text = 'Santosh', org = (50, 300), fontFace = font, fontScale =3, color = (255, 255, 255), thickness = 15)
    return blank_img
```

```python
def display_gray(img):
    fig = plt.figure(figsize=(6,6))
    ax = fig.add_subplot(111)
    ax.imshow(img, cmap = 'gray')
```

```python
img = load_img()
```

```python
display_gray(img)
```

```python
kernel = np.ones((5,5), dtype=np.uint8)
```

```python
kernel
```

```python
result = cv2.erode(img, kernel, iterations=2)
```

```python
display_gray(result)
```

# Adding noise to the image

```python
img = load_img()
```

```python
white_noise = np.random.randint(low=0, high=2, size = (600, 600))
```

```python
white_noise
```

```python
display_gray(white_noise)
```

```python
white_noise = white_noise * 255
```

```python
display_gray(white_noise)
```

```python
noise_img = white_noise + img
```

```python
display_gray(noise_img)
```

```python
opening = cv2.morphologyEx(noise_img, cv2.MORPH_OPEN, kernel)
```

```python
display_gray(opening)
```

```python
display_gray(img)
```

```python
img6 = load_img()
```

```python
black_noise = np.random.randint(low=0, high=2, size=(600,600))
```

```python
black_noise = black_noise * -255
```

```python
black_noise
```

```python
black_noise_img = img6 + black_noise
```

```python
black_noise_img
```

```python
black_noise_img[black_noise_img==-255]=0
```

```python
display_gray(black_noise_img)
```

```python
closing = cv2.morphologyEx(black_noise_img, cv2.MORPH_CLOSE, kernel)
```

```python
display_gray(closing)
```

```python
gradient = cv2.morphologyEx(img, cv2.MORPH_GRADIENT, kernel)
```

```python
display_gray(gradient)
```

# Gradients

```python
img7 = cv2.imread('.\\DATA\\sudoku.jpg', 0)
```

```python
display_gray(img7)
```

```python
sobelx = cv2.Sobel(img7, cv2.CV_64F, 1, 0, ksize=5)
```

```python
display_gray(sobelx)
```

```python
sobely = cv2.Sobel(img7, cv2.CV_64F, 0, 1, ksize=5)
```

```python
display_gray(sobely)
```

```python
laplacian = cv2.Laplacian(img7, cv2.CV_64F)
```

```python
display_gray(laplacian)
```

```python
blended_img = cv2.addWeighted(src1=sobelx, alpha=0.5, src2=sobely, beta=0.5, gamma=0)
```

```python
display_gray(blended_img)
```

```python
ret, thresh2 = cv2.threshold(img7, 100, 255, cv2.THRESH_BINARY)
```

```python
display_gray(thresh2)
```

```python
kernel1 = np.ones((4,4),np.uint8)
```

```python
gradient1= cv2.morphologyEx(blended_img, cv2.MORPH_GRADIENT, kernel1)
```

```python
display_gray(gradient1)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
img = cv2.imread('..\\DATA\\sammy.jpg')
type(img)
```

```python
plt.imshow(img)
```

```python
edges = cv2.Canny(image = img, threshold1 = 127, threshold2 = 127)
```

```python
plt.imshow(edges)
```

```python
# How to set threshold values
median = np.median(img)
print(median)
```

```python
# Lower threshold to either 0 or 70% of the median value
lower = int(max(0, 0.7*median)) 
# Upper threshold to either 255 or 130% of the median value which ever is smaller
upper = int(min(255, 1.3*median))
```

```python
edges = cv2.Canny(image = img, threshold1 = lower, threshold2 = upper)
plt.imshow(edges)
```

```python
edges = cv2.Canny(image = img, threshold1 = lower, threshold2 = upper + 80)
plt.imshow(edges)
```

```python
blurred_img = cv2.blur(img, ksize=(3, 3))
```

```python
edges = cv2.Canny(image= blurred_img, threshold1 = lower, threshold2 = upper+50)
plt.imshow(edges)
```
